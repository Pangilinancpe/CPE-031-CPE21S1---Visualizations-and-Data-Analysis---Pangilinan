3. Remove unneccessary columns and make sure numerical columns only contain numbers.

```python
df.columns
```

```python
df = df.drop(columns=['Transaction ID', 'Payment Method', 'Location', 'Transaction Date'])

df.head()
```

```python
df.dtypes ## Checking the datatype for each column
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors='coerce')
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors='coerce')
df["Total Spent"] = pd.to_numeric(df["Total Spent"], errors='coerce')

df.head()
```

```python
df.dtypes ## Checking the datatype for each column
```

QUESTION: What happened to the object columns?




the columns originally typed as object are now converted into float64

4. Remove missing data to ensure all that the final dataset is clean

```python
df.isna().sum()
```

```python
df = df.dropna()
```

```python
df.isna().sum()
```

```python
df.head()
```

EXPLAIN the process of data cleaning done in the procedure.

since the df.isna().sum() was called, it was to detect the missing values across the columns, then along df.dropna() to verify that there are no missing values remained before cleaning the dataset
